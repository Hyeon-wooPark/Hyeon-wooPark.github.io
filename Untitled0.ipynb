```python
import re
import hashlib
import uuid
from cryptography.fernet import Fernet
from faker import Faker

fake = Faker("ko_KR")

# 관리자 키 (실서비스는 환경변수/키관리시스템에 보관!)
MASTER_KEY = Fernet.generate_key()
fernet = Fernet(MASTER_KEY)

# 메모리 토큰스토어 (DB로 교체 가능)
token_store = {}
```

```python
PATTERNS = {
    "rrn": re.compile(r"\b\d{6}-\d{7}\b"),          # 주민번호
    "phone": re.compile(r"\b01[016789]-\d{3,4}-\d{4}\b"),
    "name": re.compile(r"[가-힣]{2,4}")             # 단순 이름 패턴 (데모용)
}
```

```python
def mask_with_fake(value, value_type):
    if value_type == "name":
        return fake.name()
    if value_type == "phone":
        return fake.phone_number()
    if value_type == "rrn":
        return "****-*******"
    return "***"
```

```python
def tokenize(value):
    token = str(uuid.uuid4())
    encrypted = fernet.encrypt(value.encode())
    token_store[token] = encrypted
    return token
```

```python
def hash_value(value):
    return hashlib.sha256(value.encode()).hexdigest()
```

```python
def pseudonymize(text):
    results = []

    for t, pattern in PATTERNS.items():
        for m in pattern.finditer(text):
            original = m.group()

            # 전략 예시
            if t == "rrn":
                strategy = {
                    "type": "token",
                    "pseudonym": tokenize(original)
                }
            elif t == "phone":
                strategy = {
                    "type": "hash",
                    "pseudonym": hash_value(original)
                }
            elif t == "name":
                strategy = {
                    "type": "replace",
                    "pseudonym": mask_with_fake(original, "name")
                }

            results.append({
                "type": t,
                "original": original,
                "strategy": strategy
            })

    return results
```

```python
def admin_restore(token, is_admin=False):
    if not is_admin:
        raise PermissionError("관리자 권한이 필요합니다.")

    encrypted = token_store.get(token)
    if not encrypted:
        return None

    return fernet.decrypt(encrypted).decode()
```
